```python
import sqlite3
import pandas as pd

con = sqlite3.connect("../data/raw/ipl.db")

def q(sql):
    return pd.read_sql(sql, con)
```

```python
# command 1

q("SELECT COUNT(*) FROM matches;")
```

```python
# Command 7 — COUNT and AS
q("""
SELECT COUNT(*) AS total_matches
FROM matches;
""")
# My observation:
# COUNT(*) counts the number of rows in the matches table.
```

```python
# Command 9 — GROUP BY
q("""
SELECT venue, COUNT(*) AS matches_hosted
FROM matches
GROUP BY venue
ORDER BY matches_hosted DESC;
""")
# My observation:
# GROUP BY creates one group for each venue and COUNT(*)
# counts the matches in each group.
```
