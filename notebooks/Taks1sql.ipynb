```python
import pandas as pd 
import sqlite3
  
con = sqlite3.connect(r'c:\Users\Lenovo\ipl-cricket-analytics\data\raw\ipl.db')
  
def q(sql): 
    """Run a SELECT and show the result as a table.""" 
    return pd.read_sql(sql, con)


def run_sql_file(path): 
    """Read a .sql file and execute every statement in it.""" 
    with open(path, 'r') as f: 
        script = f.read() 
    con.executescript(script) 
    con.commit() 
    print('ran', path)
```

```python
# 1. Display all data from v_ball: first 10 rows from the ball-by-ball view.
q("""
SELECT *
FROM v_ball
LIMIT 10;
""")
```

```python
# 2. Display all innings data: match-wise runs, wickets, and legal balls.
q("""
SELECT *
FROM v_innings
LIMIT 10;
""")
```

```python
# 3. Find the highest-scoring innings: top 10 innings by runs.
q("""
SELECT match_id,
       innings,
       batting_team,
       runs
FROM v_innings
ORDER BY runs DESC
LIMIT 10;
""")
```

```python
# 4. Find average innings score.
q("""
SELECT ROUND(AVG(runs), 2) AS average_score
FROM v_innings;
""")
```

```python
# 5. Find total runs scored by each team.
q("""
SELECT batting_team,
       SUM(runs) AS total_runs
FROM v_innings
GROUP BY batting_team
ORDER BY total_runs DESC;
""")
```

```python
# 6. Find average score of each team.
q("""
SELECT batting_team,
       ROUND(AVG(runs), 2) AS average_score
FROM v_innings
GROUP BY batting_team
ORDER BY average_score DESC;
""")
```

```python
# 7. Find teams with average score above 160.
q("""
SELECT batting_team,
       ROUND(AVG(runs), 2) AS average_score
FROM v_innings
GROUP BY batting_team
HAVING AVG(runs) > 160
ORDER BY average_score DESC;
""")
```

```python
# 8. Find the highest-scoring team innings.
q("""
SELECT batting_team,
       MAX(runs) AS highest_score
FROM v_innings
GROUP BY batting_team
ORDER BY highest_score DESC;
""")
```

```python
# 9. Count innings played by each team.
q("""
SELECT batting_team,
       COUNT(*) AS innings_count
FROM v_innings
GROUP BY batting_team
ORDER BY innings_count DESC;
""")
```

```python
# 10. Find matches where the first innings score was above 200.
q("""
SELECT match_id,
       first_innings_runs,
       bat_first_team,
       match_winner
FROM v_match_totals
WHERE first_innings_runs > 200
ORDER BY first_innings_runs DESC;
""")
```

```python
# 11. Find the biggest first-innings score.
q("""
SELECT MAX(first_innings_runs) AS highest_first_innings_score
FROM v_match_totals;
""")
```

```python
# 12. Count wins by team.
q("""
SELECT match_winner,
       COUNT(*) AS wins
FROM v_match_totals
GROUP BY match_winner
ORDER BY wins DESC;
""")
```

```python
# 13. Calculate chase-win percentage.
# chase_won = 1 means the chasing team won; 0 means it lost.
q("""
SELECT ROUND(100.0 * AVG(chase_won), 2) AS chase_win_percentage
FROM v_match_totals;
""")
```

```python
# 14. Find the best venues for chasing, considering venues with at least 30 matches.
# v_matches_clean provides venue_clean for the match IDs in v_match_totals.
q("""
SELECT mc.venue_clean,
       COUNT(*) AS matches,
       ROUND(100.0 * AVG(mt.chase_won), 2) AS chase_win_percentage
FROM v_match_totals mt
JOIN v_matches_clean mc
  ON mc.match_id = mt.match_id
GROUP BY mc.venue_clean
HAVING COUNT(*) >= 30
ORDER BY chase_win_percentage DESC
LIMIT 10;
""")
```

```python
# 15. Find runs scored in each innings phase.
q("""
SELECT phase,
       SUM(total_runs) AS total_runs
FROM v_ball
GROUP BY phase
ORDER BY total_runs DESC;
""")
```
